```python
import pandas as pd
import os

all_dfs = []
folder_path = './raw'
files_to_combine = [
    'linol_tracks.csv',
    'grass_tracks.csv',
    'rub_tracks.csv',
    'cell_tracks.csv'
]

for filename in files_to_combine:
    file_path = os.path.join(folder_path, filename)
    
    if os.path.exists(file_path):
        print(f"Чтение: {filename}")
        df = pd.read_csv(file_path)
        all_dfs.append(df)
    else:
        print(f"Файл НЕ НАЙДЕН: {filename}")

if all_dfs:
    combined_df = pd.concat(all_dfs, axis=0, ignore_index=True)
    print(f"Объединено файлов: {len(all_dfs)}")
    print(f"Общее количество строк: {len(combined_df)}")
    
    combined_df.to_csv(os.path.join(folder_path, "all_tracks.csv"), index=False)
else:
    print("Нет данных для объединения.")
```
