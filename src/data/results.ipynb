```python
import pandas as pd
import numpy as np
import math
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
classif_data = Path('classifier/')
results = pd.read_csv(classif_data / "cls_results.csv")
surfs = ['linoleum', 'grass', 'rubber_carpet', 'cell_carpet']

def similarityPercent(df):
    comparison_mask = df['surface'] == df['cls_raw']
    percent = comparison_mask.mean() * 100
    return percent

for surf in surfs:
    accuracy = similarityPercent(results[results['surface'] == surf])
    print(f'Surface: {surf}, accuracy: {accuracy}')

total_accuracy = similarityPercent(results)
print(f'Total accuracy: {total_accuracy}')
```

```python
classif_data = Path('classifier/')

surfs = ['linoleum', 'grass', 'rubber_carpet', 'cell_carpet']
tracks = ['straight', 'square', 'hourglass']

def similarityPercent(df):
    comparison_mask = df['surface'] == df['cls_raw']
    percent = comparison_mask.mean() * 100
    return percent

results = pd.read_csv(classif_data / f"cls_results_all_tracks.csv")
for surf in surfs:
    res_by_surf = results[results['surface'] == surf]
    print(f'Surface: {surf}')
    for track in tracks:
        accuracy = similarityPercent(res_by_surf[res_by_surf['motion_type'] == track])
        print(f'Track: {track}, accuracy, %: {accuracy:.2f}')
    print('-----')
```

```python
import pandas as pd
from pathlib import Path

classif_data = Path('classifier/')
file_path = classif_data / "cls_results_for_table.csv"

results = pd.read_csv(file_path)

results['accuracy_val'] = (results['surface'] == results['cls_raw']).astype(float) * 100

pivot_table = results.pivot_table(
    values='accuracy_val', 
    index='surface', 
    columns='motion_type', 
    aggfunc='mean'
)

pivot_table['Avg_Surface'] = pivot_table.mean(axis=1)
pivot_table.loc['Avg_Track'] = pivot_table.mean(axis=0)

print("Accuracy Table (%, by surface and track):")
print("-" * 50)
print(pivot_table.round(2))
```
